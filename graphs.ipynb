Check if there are missing values in the label column

```python
import pandas as pd

# Load the dataset
file_path = "Global.csv" 
df = pd.read_csv(file_path)

# Check for missing values in the 'Label' column
missing_values = df['Label'].isnull().sum()

# Get unique values in the 'Label' column
unique_values = df['Label'].unique()

# Display results
print(f"Missing values in 'Label' column: {missing_values}")
print(f"Unique values in 'Label' column: {unique_values}")
```

Bar Chart Malicious - Benign

```python
import matplotlib.pyplot as plt

# Count the occurrences of each label
label_counts = df['Label'].value_counts()

colors = ['blue' if label == 'Benign' else 'red' for label in label_counts.index]

# Create a bar chart
plt.figure(figsize=(6, 4))
plt.bar(label_counts.index, label_counts.values, color=colors)

# Add labels and title
plt.xlabel("Class")
plt.ylabel("Number of Samples")
plt.title("Distribution of Benign and Malicious Samples")
plt.xticks(rotation=0)  # Keep labels horizontal
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Show the plot
plt.show()
```

Check unique values in the "predicted" column

```python
# Check unique values in the "Predicted" column
unique_values = df['predicted'].unique()
print("Unique values in the 'Predicted' column:", unique_values)

# Check if there are any missing values
missing_values = df['predicted'].isnull().sum()
print("Missing values in the 'Predicted' column:", missing_values)

# Count the number of records for each network slice
slice_counts = df['predicted'].value_counts()
print("\nNumber of records for each slice:")
for value, count in slice_counts.items():
    print(f"{value}: {count} records")

```

```python

predicted_counts = df["predicted"].value_counts()

colors = ['blue', 'red', 'green']  # Adjust colors if necessary

# Bar Chart
plt.figure(figsize=(8, 5))
plt.bar(predicted_counts.index, predicted_counts.values, color=colors)
plt.xlabel("Predicted Labels")
plt.ylabel("Count")
plt.title("Distribution of Predicted Labels")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()

# Pie Chart
plt.figure(figsize=(6, 6))
plt.pie(predicted_counts, labels=predicted_counts.index, autopct="%1.1f%%", colors=colors, startangle=140)
plt.title("Predicted Labels Distribution")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the original dataset (Global.csv)
df = pd.read_csv('Global.csv')

# Ensure that 'DstPkts' and 'Label' exist
if 'DstPkts' in df.columns and 'Label' in df.columns:
    # Convert Label to categorical (if not already)
    df['Label'] = df['Label'].astype(str)

    # Plot KDE
    plt.figure(figsize=(10, 6))
    sns.kdeplot(df[df['Label'] == 'Benign']['DstPkts'], fill=True, color='blue', label='Benign', alpha=0.5)
    sns.kdeplot(df[df['Label'] == 'Malicious']['DstPkts'], fill=True, color='red', label='Malicious', alpha=0.5)

    plt.title('Kernel Density Estimation (KDE) Plot of DstPkts')
    plt.xlabel('Destination Packets (DstPkts)')
    plt.ylabel('Density')
    plt.legend()
    plt.show()
else:
    print("DstPkts or Label column not found in dataset!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the original dataset (Global.csv)
df = pd.read_csv('Global.csv')

# Ensure that 'State' and 'Label' exist
if 'State' in df.columns and 'Label' in df.columns:
    # Convert Label and State to categorical (if not already)
    df['Label'] = df['Label'].astype(str)
    df['State'] = df['State'].astype(str)

    # Plot KDE of State occurrences (countplot, since State is categorical)
    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, x='State', hue='Label', palette={'Benign': 'blue', 'Malicious': 'red'})

    plt.title('State Feature Distribution by Traffic Type')
    plt.xlabel('State')
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.legend(title='Label')
    plt.tight_layout()
    plt.show()
else:
    print("State or Label column not found in dataset!")

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the original dataset (Global.csv)
df = pd.read_csv('Global.csv')

# Ensure that 'sMeanPktSz' and 'Label' exist
if 'sMeanPktSz' in df.columns and 'Label' in df.columns:
    # Convert Label to categorical (if not already)
    df['Label'] = df['Label'].astype(str)

    # Define color mapping for Label categories
    color_palette = {'Benign': 'blue', 'Malicious': 'red'}

    # Plot Boxplot
    plt.figure(figsize=(8, 6))
    sns.boxplot(x='Label', y='sMeanPktSz', hue='Label', data=df, palette=color_palette, legend=False)

    plt.title('Boxplot of sMeanPktSz (Mean Packet Size)')
    plt.xlabel('Traffic Type (Benign vs Malicious)')
    plt.ylabel('Mean Packet Size (sMeanPktSz)')
    plt.grid(True)
    plt.show()
else:
    print("sMeanPktSz or Label column not found in dataset!")

```
